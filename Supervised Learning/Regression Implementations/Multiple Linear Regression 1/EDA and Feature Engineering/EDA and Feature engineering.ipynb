```python
#import the libaries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("Algerian_forest_fires_dataset.csv",header=1)
#header =i, tells pandas to use the second row as the header, which contains the column names. By default, pandas uses the first row (index 0) as the header, but in this case, the actual column names are in the second row (index 1).
df.head()
```

```python
df.info()
```

```python
#data cleaning
df[df.isnull().any(axis=1)]
```

##### The dataset is comverted into 2 sets based on region from 122th index, we can make a column based on the region, such that BEJAIA REGION as 0 and SIDI BEL ABBES REGION as 1

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
df.info()
```

```python
df["Region"]=df["Region"].astype(int)
df.head()
```

```python
df.isnull().sum()
```

```python
# Removing the null values and resetting the index to adjust the gap
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
#the names of the columns are not in the right format, so we will change them to the right format
```

```python
df.columns=df.columns.str.strip()
df.columns
```

### Changing the required columns as integer data type

```python
df[["month","day","year","Temperature", "RH", "Ws"]]=df[["month","day","year","Temperature", "RH", "Ws"]].astype(int)
df.info()
```

### Changing the other columns to float data type

```python
df[[ 'Rain', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']]=df[[ 'Rain', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype(float)
df.info()
```

```python
df.describe()
```

```python
#Saving the cleaned dataset into a new csv file
df.to_csv("cleaned_algerian_forest_fires_dataset.csv",index=False)
```

# Exploratory Data Analysis (EDA)

```python
# dropping date, month,year, as they are unnecessary for our analysis and model building
df_copy=df.drop(columns=["month","day","year"],axis=1)
df_copy.head()
```

```python
# Encoding of the categories in Classes
df_copy["Classes"] = df_copy["Classes"].astype(str).str.strip()
df_copy["Classes"] = np.where(
    df_copy["Classes"].str.contains("not fire"),
    0,
    1
)
```

```python
# categories in classes column
df_copy["Classes"].value_counts()
```

```python
# plot density plot for all features
plt.style.use('seaborn-v0_8')
df_copy.hist(figsize=(15,10),bins=50)
plt.show()
```

```python
# % for pie chart
percentage=df_copy["Classes"].value_counts(normalize=True)*100
```

```python
# plotting piechart
Classlabels=["Fire","Not Fire"]
plt.figure(figsize=(5,5))
plt.pie(percentage,labels=Classlabels,autopct="%1.1f%%")
plt.title("Distribution of Fire and Not Fire Classes")
```

```python
# Correlation
df_copy.corr()
```

```python
# Encoding of the categories in Classes
df["Classes"] = df["Classes"].astype(str).str.strip()
df["Classes"] = np.where(
    df["Classes"].str.contains("not fire"),
    "not fire",
    "fire"
)
```

```python
# Monthly Fire Analysis
dftemp=df.loc[df['Region']==1]
plt.subplots(figsize=(15,10))
sns.set_style("whitegrid")
sns.countplot(x='month',data=df,hue='Classes')
plt.title("Monthly Fire Analysis for Sidi bel Region",weight="bold")
plt.xlabel("Month",weight="bold")
plt.ylabel("Count",weight="bold")
```

```python
# Monthly Fire Analysis
dftemp=df.loc[df['Region']==0]
plt.subplots(figsize=(15,10))
sns.set_style("whitegrid")
sns.countplot(x='month',data=df,hue='Classes')
plt.title("Monthly Fire Analysis for bejaia Region",weight="bold")
plt.xlabel("Month",weight="bold")
plt.ylabel("Count",weight="bold")
```

##### Its observed that August and September had the most number of forest fires for both the regions, most of the fires happened in August and very high fires happened in only 3 months - June,July and August and less fires during September
