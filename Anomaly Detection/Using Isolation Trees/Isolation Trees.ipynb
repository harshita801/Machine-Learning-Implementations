```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.ensemble import IsolationForest
from sklearn.cluster import DBSCAN
```

```python
X, _ = make_blobs(
    n_samples=300,
    centers=1,
    n_features=2,
    cluster_std=1.0,
    random_state=42
)

outliers = np.array([
    [8, 8],
    [9, -7],
    [-8, 7]
])

X = np.vstack([X, outliers])
```

```python
plt.scatter(X[:,0], X[:,1])
```

```python
clf=IsolationForest(contamination=0.02)
clf.fit(X)
predictions=clf.predict(X)
```

```python
predictions
```

```python
index=np.where(predictions==-1)
index
```

```python
plt.scatter(X[:,0], X[:,1])
plt.scatter(X[index,0], X[index,1], edgecolor='red')
```
